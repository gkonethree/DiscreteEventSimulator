```python
import subprocess
import matplotlib.pyplot as plt
```

```python
def run_simulation(num_users):
    stats = {}
    # Run the simulator with the specified number of users
    result = subprocess.run(["./simulate", str(num_users)], capture_output=True, text=True).stdout
    # print(result)
    # for line in result.split("\n")[:-1]:
    #     tmp = line.split(":")
    #     print(tmp)
    #     stats[tmp[0].strip()] = float(tmp[1].strip())
    res_time = result.split("\n")[0].split(":")[1].strip()
    stats["Average Cycle Time"] = float(res_time)
    rsys = result.split("\n")[1].split(":")[1].strip()
    stats["Average Rsys Time"] = float(rsys)
    tgpt= result.split("\n")[2].split(":")[1].strip()
    stats["Throughput"] = float(tgpt)
    stats["avg server utilization"] = float(result.split("\n")[7].split(":")[1].strip()[:-1])
    stats["frac dropped"] = float(result.split("\n")[5].split(":")[1].strip())
    return stats
```

```python
restime = []
rsys = []
tgpt = []
utilization = []
dropped = []

users=[50,70,90,100,120,140,150,155,160,170,180,190]
for user in users:
    restime.append(run_simulation(user)["Average Cycle Time"])
    tgpt.append(run_simulation(user)["Throughput"])
    rsys.append(run_simulation(user)["Average Rsys Time"])
    utilization.append(run_simulation(user)["avg server utilization"])
    dropped.append(run_simulation(user)["frac dropped"])

tgpt = [x * 100 for x in tgpt]
```

```python
plt.plot(users, rsys, marker='o')
plt.xlabel("Number of Users")
plt.ylabel("Average Rsys Time")
plt.title("Performance of the System")
plt.grid()
plt.savefig("rsys_user.png")
plt.clf()

plt.plot(users, tgpt, marker='o')
plt.xlabel("Number of Users")
plt.ylabel("Throughput")
plt.title("Throughput of the System")
plt.grid()
plt.savefig("throughput_users.png")
plt.clf()

    
plt.plot(users,utilization, marker='o')
plt.xlabel("Users")
plt.ylabel("Server Utilization")
plt.title("Server Utilization vs Users")
plt.grid()
plt.savefig("utilization_users.png")
plt.clf()

plt.plot(users,dropped, marker='o')
plt.xlabel("Users")
plt.ylabel("Fraction of Requests Dropped")
plt.title("Fraction of Requests Dropped vs Users")
plt.grid()
plt.savefig("dropped_users.png")
plt.clf()
```
